## Bài 1 · Index và slicing
Từ `text`, lấy ký tự đầu, ký tự cuối, 6 ký tự đầu và chuỗi đảo ngược.

**Kết quả:** `P`, `T`, `Python`, `TIH nohtyP`.

```python
text = "Python HIT"

first_char = text[0]
last_char = text[-1]
first_six = text[0:6]
reversed_text = text[::-1]

print(first_char, last_char, first_six, reversed_text, sep="\n")
```

## Bài 2 · Chuẩn hóa tên
Bỏ khoảng trắng thừa hai đầu và giữa các từ, chuyển sang chữ thường. Đếm số từ trong tên đã chuẩn hóa.

**Gợi ý:** `strip()`, `lower()`, `split()`, `" ".join(...)`, `len()`.

**Kết quả:** `nguyen van an`, số từ `3`.

```python
raw_name = "  NGUYEN   Van   An  "

clean_name =" ".join(raw_name.lower().strip().split())
word_count = len(clean_name.split())

print(clean_name)
print(word_count)
```

## Bài 3 · Thêm, sửa, xóa list
Thực hiện lần lượt:
1. Thêm điểm 10 vào cuối.
2. Sửa phần tử ở index 1 thành 9.
3. Xóa điểm 6 bằng `remove()`.

**Kết quả:** `[7, 9, 10]`.

**Lưu ý:** `append()` và `remove()` sửa list tại chỗ; không gán kết quả của chúng trở lại biến list.

```python
scores_b3 = [7, 8, 6]
scores_b3.append(10)
scores_b3.insert(1,9)
scores_b3.remove(6)
scores_b3.pop(2)
print(scores_b3)
```

## Bài 4 · Lọc và tổng hợp điểm
Dùng `for` kết hợp `if`, tạo list `passed_scores` chứa các điểm từ 5 trở lên. Tính trung bình của **toàn bộ** điểm vào `average_score`.

**Kết quả:** `[8, 5, 9]`, trung bình `6.0`.

Thử thêm với list rỗng: đặt trung bình là `None` và in `Chưa có điểm`, tránh chia cho 0.

```python
scores_b4 = [4, 8, 5, 9, 4]
passed_scores = []
average_score = 0

for i in scores_b4:
    if i >= 5:
        passed_scores.append(i)

average_score = sum(passed_scores) / len(passed_scores)
print(passed_scores)
print(average_score)
```

## Bài 5 · Tuple và set
- Tách `size` thành `width`, `height`, rồi tính diện tích.
- Từ `names`, tạo set các tên không trùng và đếm số tên khác nhau.

**Kết quả:** diện tích `60`; set chứa `An`, `Binh`, `Chi`; số tên `3`.

Set không đảm bảo thứ tự hiển thị. Có thể dùng `sorted()` để in kết quả ổn định.

```python
size = (12, 5)
names = ["An", "Binh", "An", "Chi", "Binh"]

width = size[0]
height = size[1]
area = width * height
unique_names = set(names)
unique_count = len(unique_names)

print(area)
print(unique_names)
print(unique_count)
```

## Bài 6 · Cập nhật dict
Sửa điểm thành 9, thêm nhóm `HIT`, xóa khóa `email`. Đọc khóa `phone` bằng `get()`, dùng giá trị mặc định `Chưa có`.

**Kết quả:** `student` còn `name`, `score`, `group` với giá trị `An`, `9`, `HIT`; `phone` là `Chưa có`.

```python
student = {"name": "An", "score": 8, "email": "an@example.com"}
for x, y in student.items():
    print(x,y)
student.get("class", "khong co")
student.pop("email")
```

```python
student = {"name": "An", "score": 8, "email": "an@example.com"}

student["score"] = 9
student["group"] = "HIT"
student.pop("email")
phone = student.get("Phone","chua co")

print(student)
print(phone)
```

## Bài 7 · Đếm từ
Chuyển văn bản thành chữ thường, tách từ rồi dùng vòng lặp để đếm số lần xuất hiện vào dict `counts`.

**Gợi ý:** `counts.get(word, 0)` lấy số lần đã đếm, mặc định là 0 khi chưa có từ.

**Kết quả:** `{'python': 3, 'hit': 2, 'code': 1}`.

Giả sử văn bản chỉ gồm các từ và khoảng trắng, chưa cần xử lý dấu câu.

```python
text_b7 = "Python HIT python code HIT Python"
text_b7.lower()
counts = {}

for word in text_b7:
    counts[word] = counts.get(word, 0) + 1
print(counts)
```
