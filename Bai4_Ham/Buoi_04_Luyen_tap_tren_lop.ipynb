## Bài 1 — Máy tính diện tích
Viết `dien_tich(dai, rong)` để **trả về** diện tích hình chữ nhật.

**Kết quả mẫu:** `dien_tich(5, 3)` → `15`; `dien_tich(2.5, 4)` → `10.0`.

```python
def dien_tich(dai, rong):
    return (dai + rong) * 2

print(dien_tich(5, 3))
print(dien_tich(2.5, 4))
```

Hoàn thiện `gap_doi(x)` để kết quả có thể **dùng tiếp cho phép tính**.

## Bài 3 — Lời chào mặc định
Viết `chao(ten, loi_chao="Xin chao")` để trả về chuỗi lời chào.

**Kết quả mẫu:**
- `chao("An")` → `Xin chao, An!`
- `chao("An", "Hello")` → `Hello, An!`

```python
def chao(ten, loi_chao="Xin chao"):
    return print(loi_chao,{ten})

print(chao("An"))
print(chao("An", "Hello"))
```

## Bài 5 — Chuẩn hóa họ tên
Viết `chuan_hoa_ten(ten)` để bỏ khoảng trắng thừa và viết hoa chữ cái đầu mỗi từ.

**Kết quả mẫu:**
- `"  nGUYEN   van an  "` → `Nguyen Van An`
- `"   tRAN   THI   MAI "` → `Tran Thi Mai`

*Gợi ý:* `split()`, `join()`, `title()`.

```python
def chuan_hoa_ten(ten):
    ten = " ".join(ten.split().title())
    return ten

print(chuan_hoa_ten("  nGUYEN   van an  "))
print(chuan_hoa_ten("   tRAN   THI   MAI "))
```

## Bài 6 — Báo cáo điểm bằng **3 hàm**
Tự hoàn thiện:
1. `tinh_tb(scores)`: trả về điểm trung bình.
2. `xep_loai(tb)`: `>=8` → `Giỏi`, `>=5` → `Đạt`, còn lại → `Cần cố gắng`.
3. `tao_bao_cao(scores)`: **gọi hai hàm trên** rồi trả về dòng báo cáo.

**Kết quả mẫu:** `[8, 4, 9, 6]` → `Điểm TB: 6.75 | Đạt`.

*Dùng `:.2f` để hiển thị hai chữ số thập phân.*

```python
def tinh_tb(scores):
    tong= 0
    for score in scores:
        tong += score
    tb = tong/len(scores)
    return tb


def xep_loai(tb):
    if tb >= 8:
        return "Gioi"
    elif tb >= 5:
        return "Dat"
    else:
        return "Can co gang"
    
def tao_bao_cao(scores):
    tb = tinh_tb(scores)
    loai = xep_loai(tb)
    return f"Diem TB: {tb:.2f} | {loai}"


print(tao_bao_cao([8, 4, 9, 6]))
```

## ⭐ Bài thêm — Scope: không làm đổi biến gốc
Viết `cong_diem(diem, thuong=1)` để **trả về** điểm mới, tối đa là `10`. Không sửa biến `diem_goc` bên ngoài hàm.

**Kết quả mẫu:** lần lượt in `9`, `10`, `7`.

```python
diem_goc = 7

def cong_diem(diem, thuong=1):
    diem_moi = diem + thuong
    if diem_moi > 10:
        diem_moi = 10
    return diem_moi

print(cong_diem(diem_goc, 2))
print(cong_diem(diem_goc, 5))
print(diem_goc)
```
